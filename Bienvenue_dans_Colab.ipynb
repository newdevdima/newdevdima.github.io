```
import numpy as np
import IPython.display as display
from matplotlib import pyplot as plt
import io
import base64

ys = 200 + np.random.randn(100)
x = [x for x in range(len(ys))]

fig = plt.figure(figsize=(4, 3), facecolor='w')
plt.plot(x, ys, '-')
plt.fill_between(x, ys, 195, where=(ys > 195), facecolor='g', alpha=0.6)
plt.title("Sample Visualization", fontsize=10)

data = io.BytesIO()
plt.savefig(data)
image = F"data:image/png;base64,{base64.b64encode(data.getvalue()).decode()}"
alt = "Sample Visualization"
display.display(display.Markdown(F"""![{alt}]({image})"""))
plt.close(fig)
```

```
# NumPy - Calcul scientifique et opérations matrices
import numpy as np
np.random.seed(42)

# Pandas - Manipulation et analyse données
import pandas as pd
pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:.3f}'.format)

# Matplotlib - Visualisation bas niveau
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.gridspec import GridSpec
plt.style.use('seaborn-v0_8-darkgrid')

# Seaborn - Visualisation statistique haut niveau
import seaborn as sns
sns.set_palette('husl')

# SciPy - Statistiques avancées
from scipy import stats
from scipy.stats import normaltest, kurtosis, skew, zscore
from scipy.cluster.hierarchy import dendrogram, linkage

# Scikit-learn - Machine Learning et preprocessing
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.ensemble import IsolationForest
from sklearn.metrics import silhouette_score

# Warnings
import warnings
warnings.filterwarnings('ignore')

# Affichage
%matplotlib inline
plt.rcParams['figure.figsize'] = (14, 8)
plt.rcParams['font.size'] = 10

print("✓ Toutes les librairies importées avec succès!")
```

```
# Chargement du dataset
df = pd.read_csv('credit_risk_dataset.csv', sep=';')

print("="*80)
print("INFORMATION GÉNÉRALE DU DATASET")
print("="*80)
print(f"\n📊 Dimensions: {df.shape[0]} lignes × {df.shape[1]} colonnes")
print(f"\n💾 Taille en mémoire: {df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")

# Affichage des premières lignes
print("\n📋 Aperçu des 5 premières lignes:")
display(df.head())

# Types de données
print("\n🔍 Types de données:")
print(df.dtypes)

# Informations détaillées
print("\n📝 Informations détaillées:")
df.info()
```

```
print("="*80)
print("ANALYSE QUALITÉ DES DONNÉES")
print("="*80)

# Données manquantes
missing_data = df.isnull().sum()
missing_percent = (missing_data / len(df)) * 100
missing_df = pd.DataFrame({'Manquantes': missing_data, 'Pourcentage': missing_percent})
missing_df = missing_df[missing_df['Manquantes'] > 0].sort_values('Pourcentage', ascending=False)

if len(missing_df) > 0:
    print("\n❌ Données manquantes détectées:")
    display(missing_df)
else:
    print("\n✓ Aucune donnée manquante")

# Doublons
duplicates = df.duplicated().sum()
print(f"\n🔄 Doublons: {duplicates} ({duplicates/len(df)*100:.2f}%)")

# Variables numériques vs catégorielles
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = df.select_dtypes(include=['object']).columns.tolist()

print(f"\n📊 Variables numériques ({len(numeric_cols)}): {', '.join(numeric_cols)}")
print(f"\n🏷️  Variables catégorielles ({len(categorical_cols)}): {', '.join(categorical_cols)}")
```

```
print("="*80)
print("STATISTIQUES DESCRIPTIVES UNIVARIÉES")
print("="*80)

# Statistiques de base (Pandas)
print("\n📊 Statistiques descriptives:")
display(df[numeric_cols].describe())

# Statistiques avancées avec NumPy
print("\n🔬 Statistiques avancées (NumPy):")
stats_advanced = pd.DataFrame({
    'Variable': numeric_cols,
    'Skewness': [skew(df[col].dropna()) for col in numeric_cols],
    'Kurtosis': [kurtosis(df[col].dropna()) for col in numeric_cols],
    'CV(%)': [np.std(df[col].dropna())/np.mean(df[col].dropna())*100 for col in numeric_cols],
    'IQR': [df[col].quantile(0.75) - df[col].quantile(0.25) for col in numeric_cols]
})
display(stats_advanced)

# Variables catégorielles
print("\n🏷️  Distributions catégorielles:")
for col in categorical_cols:
    print(f"\n{col}:")
    print(df[col].value_counts())
    print(f"  → Modalités uniques: {df[col].nunique()}")
```

```
# Visualisation des distributions univariées
fig = plt.figure(figsize=(16, 12))
gs = GridSpec(4, 3, figure=fig, hspace=0.3, wspace=0.3)

for idx, col in enumerate(numeric_cols):
    # Histogramme + KDE
    ax1 = fig.add_subplot(gs[idx, 0:2])
    sns.histplot(data=df, x=col, kde=True, stat='density', ax=ax1, color='steelblue', alpha=0.6)
    ax1.set_title(f'{col} - Distribution', fontsize=11, fontweight='bold')
    ax1.axvline(df[col].mean(), color='red', linestyle='--', label=f'Mean={df[col].mean():.2f}')
    ax1.axvline(df[col].median(), color='green', linestyle='--', label=f'Median={df[col].median():.2f}')
    ax1.legend(fontsize=8)

    # Boxplot
    ax2 = fig.add_subplot(gs[idx, 2])
    sns.boxplot(y=col, data=df, ax=ax2, color='lightcoral')
    ax2.set_title(f'{col} - Boxplot', fontsize=11, fontweight='bold')

plt.suptitle('ANALYSE UNIVARIÉE - Distributions des Variables Numériques',
             fontsize=14, fontweight='bold', y=0.995)
plt.show()

print("✓ Visualisations univariées complètes")
```

```
print("="*80)
print("ANALYSE BIVARIÉE - CORRÉLATIONS")
print("="*80)

# Corrélation Pearson (linéaire)
print("\n📊 Corrélation de Pearson (linéaire):")
corr_pearson = df[numeric_cols].corr(method='pearson')
display(corr_pearson)

# Corrélation Spearman (rank-based, monotonicité)
print("\n🎚️  Corrélation de Spearman (monotone):")
corr_spearman = df[numeric_cols].corr(method='spearman')
print(corr_spearman.round(3))

# Test de significativité (SciPy)
print("\n📈 Test de significativité des corrélations (SciPy):")
from scipy.stats import pearsonr, spearmanr

sig_corr = []
for i, col1 in enumerate(numeric_cols):
    for col2 in numeric_cols[i+1:]:
        tmp = df[[col1, col2]].dropna()
        r, p_value = pearsonr(tmp[col1], tmp[col2])
        if p_value < 0.05 and abs(r) > 0.1:
            sig_corr.append({
                'Variable1': col1,
                'Variable2': col2,
                'Pearson r': r,
                'p-value': p_value,
                'Significatif': '✓' if p_value < 0.05 else '✗'
            })

if sig_corr:
    sig_df = pd.DataFrame(sig_corr).sort_values('Pearson r', key=abs, ascending=False)
    display(sig_df.head(10))
```

```
print("="*80)
print("DÉTECTION DES OUTLIERS")
print("="*80)

# Méthode 1: Z-score (SciPy)
print("\n📊 Méthode 1: Z-score (|z| > 3)")
z_scores = np.abs(stats.zscore(df[numeric_cols].fillna(df[numeric_cols].mean())))
outliers_zscore = (z_scores > 3).sum(axis=0)
print(f"Outliers détectés par variable:")
for col, count in zip(numeric_cols, outliers_zscore):
    print(f"  {col}: {count} ({count/len(df)*100:.2f}%)")

# Méthode 2: IQR
print("\n📊 Méthode 2: IQR (Interquartile Range)")
outliers_iqr_count = 0
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    outliers = ((df[col] < Q1 - 1.5*IQR) | (df[col] > Q3 + 1.5*IQR)).sum()
    print(f"  {col}: {outliers} ({outliers/len(df)*100:.2f}%)")
    outliers_iqr_count += outliers

# Méthode 3: Isolation Forest (Scikit-learn)
print("\n📊 Méthode 3: Isolation Forest (ML)")
iso_forest = IsolationForest(contamination=0.05, random_state=42)
outlier_pred = iso_forest.fit_predict(df[numeric_cols].fillna(df[numeric_cols].mean()))
outliers_if = (outlier_pred == -1).sum()
print(f"Outliers détectés: {outliers_if} ({outliers_if/len(df)*100:.2f}%)")

# Visualisation
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Scatter avec outliers IQR
df['outlier_iqr'] = False
for col in ['person_income', 'loan_amnt']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    df.loc[(df[col] < Q1 - 1.5*IQR) | (df[col] > Q3 + 1.5*IQR), 'outlier_iqr'] = True

sns.scatterplot(x='person_income', y='loan_amnt', data=df,
               hue='outlier_iqr', palette=['blue', 'red'], ax=axes[0], alpha=0.6)
axes[0].set_title('Outliers détectés - Méthode IQR', fontsize=12, fontweight='bold')

# Isolation Forest
df['outlier_if'] = outlier_pred == -1
sns.scatterplot(x='person_income', y='loan_amnt', data=df,
               hue='outlier_if', palette=['blue', 'red'], ax=axes[1], alpha=0.6)
axes[1].set_title('Outliers détectés - Isolation Forest', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()

print("\n✓ Outliers détectés et visualisés")
```

```
print("="*80)
print("ANALYSE MULTIVARIÉE - PCA")
print("="*80)

# Préparation des données
X = df[numeric_cols].fillna(df[numeric_cols].mean())
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# PCA
pca = PCA()
X_pca = pca.fit_transform(X_scaled)

# Variance expliquée
var_exp = pca.explained_variance_ratio_
cumsum_var = np.cumsum(var_exp)

print("\n📊 Variance expliquée par composante:")
for i, (ve, ce) in enumerate(zip(var_exp[:5], cumsum_var[:5])):
    print(f"  PC{i+1}: {ve*100:.2f}% | Cumulée: {ce*100:.2f}%")

# Visualisation variance
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Scree plot
axes[0].plot(range(1, len(var_exp)+1), cumsum_var, 'bo-', linewidth=2, markersize=6)
axes[0].axhline(y=0.95, color='r', linestyle='--', label='95% variance')
axes[0].set_xlabel('Nombre de composantes')
axes[0].set_ylabel('Variance cumulée')
axes[0].set_title('Scree Plot - Variance Expliquée', fontweight='bold')
axes[0].grid(True, alpha=0.3)
axes[0].legend()

# Biplot (PC1 vs PC2)
loadings = pca.components_[:2].T * np.sqrt(pca.explained_variance_[:2])
axes[1].scatter(X_pca[:, 0], X_pca[:, 1], alpha=0.5, s=30, c=df['loan_status'], cmap='viridis')
for i, (feature, loading) in enumerate(zip(numeric_cols, loadings)):
    axes[1].arrow(0, 0, loading[0]*3, loading[1]*3, head_width=0.1, head_length=0.1, fc='red', ec='red')
    axes[1].text(loading[0]*3.2, loading[1]*3.2, feature, fontsize=9, ha='center')
axes[1].set_xlabel(f'PC1 ({var_exp[0]*100:.1f}%)')
axes[1].set_ylabel(f'PC2 ({var_exp[1]*100:.1f}%)')
axes[1].set_title('Biplot - PC1 vs PC2', fontweight='bold')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("\n✓ PCA complète et visualisée")
```

```
print("="*80)
print("CLUSTERING - K-MEANS")
print("="*80)

# Déterminer le nombre optimal de clusters (Elbow method + Silhouette)
inertias = []
silhouette_scores = []
K_range = range(2, 11)

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertias.append(kmeans.inertia_)
    silhouette_scores.append(silhouette_score(X_scaled, kmeans.labels_))

# Visualiser
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Elbow plot
axes[0].plot(K_range, inertias, 'bo-', linewidth=2, markersize=8)
axes[0].set_xlabel('Nombre de clusters (k)')
axes[0].set_ylabel('Inertie')
axes[0].set_title('Elbow Method', fontweight='bold')
axes[0].grid(True, alpha=0.3)

# Silhouette plot
best_k = K_range[np.argmax(silhouette_scores)]
axes[1].plot(K_range, silhouette_scores, 'go-', linewidth=2, markersize=8)
axes[1].axvline(x=best_k, color='r', linestyle='--', label=f'Optimal k={best_k}')
axes[1].set_xlabel('Nombre de clusters (k)')
axes[1].set_ylabel('Silhouette Score')
axes[1].set_title('Silhouette Analysis', fontweight='bold')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print(f"\n✓ Nombre optimal de clusters: {best_k} (Silhouette score: {max(silhouette_scores):.3f})")

# Clustering final avec k optimal
kmeans_final = KMeans(n_clusters=best_k, random_state=42, n_init=10)
clusters = kmeans_final.fit_predict(X_scaled)
df['cluster'] = clusters

print(f"\n📊 Distribution des clusters:")
print(df['cluster'].value_counts().sort_index())
```
